# Reproduce result summaries from exported outputs

This notebook loads the exported CSV files from `outputs/full_run/` and displays the main benchmark, ranking, emulation, statistical and figure outputs used in the manuscript.

```
from pathlib import Path
import pandas as pd
from IPython.display import display, Image

ROOT = Path('..')
OUT = ROOT / 'outputs' / 'full_run'
print('Output directory:', OUT.resolve())
print('Exists:', OUT.exists())
```

```
benchmark = pd.read_csv(OUT / 'benchmark_full.csv')
summary = pd.read_csv(OUT / 'summary_full.csv')
rankings = pd.read_csv(OUT / 'rankings_mase_full.csv')
stats = pd.read_csv(OUT / 'statistical_comparison_full.csv')
emulation = pd.read_csv(OUT / 'emulation_fidelity_full.csv')
emulation_summary = pd.read_csv(OUT / 'emulation_summary_full.csv')
datasets = pd.read_csv(OUT / 'dataset_manifest.csv')

print('benchmark:', benchmark.shape)
print('summary:', summary.shape)
print('rankings:', rankings.shape)
print('statistical comparisons:', stats.shape)
print('emulation:', emulation.shape)
print('datasets:', datasets.shape)
```

## Dataset manifest

```
display(datasets)
```

## Average MASE ranking

```
avg_rank = rankings.groupby('model')['rank_MASE'].mean().sort_values().reset_index(); display(avg_rank)
```

## MASE summary by dataset and model

```
display(summary[['dataset','model','MASE_mean','MASE_std','RMSE_mean','MAE_mean','inference_ms_per_sample_mean','param_count_mean']].sort_values(['dataset','MASE_mean']).head(30))
```

## Functional emulation summary

```
display(emulation_summary[['dataset','model','emul_Pearson_r_mean','emul_spectral_distance_mean','emul_autocorr_distance_mean','observable_MSE_mean','observable_Pearson_r_mean']].sort_values(['dataset','model']))
```

## Statistical comparison snapshot

```
display(stats.head(20))
```

## Main article figures

```
fig_dir = OUT / 'figures' / 'core_article'
for fig in sorted(fig_dir.glob('*.png')):
    print(fig.name)
    display(Image(filename=str(fig)))
```
